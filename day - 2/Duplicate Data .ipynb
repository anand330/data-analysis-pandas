 # duplicated()

```python
# duplicated() → duplicate rows identify karta hai.

# True → Duplicate
# False → Not duplicate / first occurrence

# Count duplicates:
# df.duplicated().sum()

# Duplicate rows:
# df[df.duplicated()]

# Specific column:
# df.duplicated("Name")

# Multiple columns:
# df.duplicated(["Name", "City"])

# All duplicate occurrences:
# df.duplicated(keep=False)
```

```python
import pandas as pd

data = {
    "Employee_ID": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        102, 104, 107, 110, 111
    ],

    "Name": [
        "Amit", "Rahul", "Neha", "Pooja", "Rohit",
        "Sneha", "Vikas", "Kiran", "Ankit", "Priya",
        "Rahul", "Pooja", "Vikas", "Priya", "Mohit"
    ],

    "Department": [
        "IT", "HR", "Finance", "Sales", "IT",
        "HR", "IT", "Finance", "Sales", "IT",
        "HR", "Sales", "IT", "IT", "Marketing"
    ],

    "City": [
        "Delhi", "Jaipur", "Mumbai", "Pune", "Delhi",
        "Jaipur", "Delhi", "Mumbai", "Pune", "Delhi",
        "Jaipur", "Pune", "Delhi", "Delhi", "Ajmer"
    ],

    "Salary": [
        35000, 42000, 48000, 50000, 45000,
        39000, 55000, 46000, 41000, 52000,
        42000, 50000, 55000, 52000, 36000
    ]
}

df = pd.DataFrame(data)
df
```

```python
# Q1. Complete DataFrame mein duplicate rows identify karo.
df.duplicated()


# Q2. Complete DataFrame mein total duplicate rows count karo.
df.duplicated().sum()


# Q3. Sirf duplicate rows select karo.
df[df.duplicated()]


# Q4. Name column ke basis par duplicate records identify karo.
df.duplicated("Name")


# Q5. Name aur Salary dono columns ke basis par duplicate records identify karo.
df.duplicated(["Name", "Salary"])


# Q6. Employee_ID ke basis par duplicate records identify karo.
df.duplicated("Employee_ID")


# Q7. Employee_ID ke basis par total duplicate records count karo.
df.duplicated("Employee_ID").sum()


# Q8. Name column ke duplicate records ko select karo.
df[df.duplicated("Name")]


# Q9. Name aur City ke basis par duplicate records select karo.
df[df.duplicated(["Name", "City"])]


# Q10. Employee_ID ke basis par saare duplicate occurrences identify karo.
# keep=False ka use karo.
df.duplicated("Employee_ID", keep=False)
```

# drop_duplicates()

```python
# drop_duplicates() is used to remove duplicate rows
# from a Pandas DataFrame.

# duplicated() → identifies duplicate rows
# drop_duplicates() → removes duplicate rows

# Syntax:
# df.drop_duplicates()
```

```python
# Complete duplicate rows
df.drop_duplicates()

# Specific column
df.drop_duplicates("Name")

# Multiple columns
df.drop_duplicates(["Name", "Salary"])

# Keep first
df.drop_duplicates(keep="first")

# Keep last
df.drop_duplicates(keep="last")

# Remove all duplicate occurrences
df.drop_duplicates(keep=False)

# Permanently update DataFrame
df.drop_duplicates(inplace=True)

```

```python
# duplicated()        → Duplicate identify
# duplicated().sum()  → Duplicate count
# df[df.duplicated()] → Duplicate rows show
# drop_duplicates()   → Duplicate remove
```
