# Number Conversion = pd.to_numeric()

```python
# pd.to_numeric() is used to convert values into numeric data.

# It can convert:
# String → Integer/Float
# Object → Numeric

# Syntax:
# pd.to_numeric(df["Column"])

# Basic numeric conversion
# pd.to_numeric(df["Salary"])

# Invalid values ko NaN mein convert karna
#pd.to_numeric(
    #df["Salary"],
    #errors="coerce" )

# Invalid value par error
#pd.to_numeric(
    #df["Salary"],
    #errors="raise")

```

```python
import pandas as pd

data = {
    "Order_ID": [
        "1001", "1002", "1003", "1004", "1005",
        "1006", "1007", "1008", "1009", "1010"
    ],

    "Product": [
        "Laptop", "Mobile", "Tablet", "Headphone", "Monitor",
        "Keyboard", "Mouse", "Printer", "Laptop", "Mobile"
    ],

    "Quantity": [
        "2", "3", "5", "2", "4",
        "Unknown", "3", "Five", "2", "4"
    ],

    "Price": [
        "55000", "25000", "30000", "5000", "22000",
        "1500", "800", "12000", "56000", "26000"
    ],

    "Discount": [
        "10", "5", "8", "None", "10",
        "5", "Invalid", "7", "10", "None"
    ],

    "Sales": [
        "110000", "75000", "150000", "10000", "88000",
        "Invalid", "2400", "84000", "112000", "104000"
    ]
}

df = pd.DataFrame(data)
df

df.dtypes

# Q1. Order_ID column ko numeric datatype mein convert karo.

df["Order_ID"]= pd.to_numeric(df["Order_ID"])

# Q2. Quantity column ko numeric mein convert karo.
# Invalid values ko NaN mein convert karo using errors="coerce".

df["Quantity"]=pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

# Q3. Discount column ko numeric mein convert karo.
# Invalid values ko NaN mein convert karo.
df["Discount"]=pd.to_numeric(
    df["Discount"],
    errors="coerce"
)

# Q4. Sales column ko numeric mein convert karo.
# Invalid values ko NaN mein convert karo.

df["Sales"]=pd.to_numeric(
    df["Sales"],
    errors="coerce"
)

# Q5. Price aur Sales dono columns ko numeric mein convert karo.
# Invalid values ko NaN mein convert karne ke liye errors="coerce" use karo.

df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)

df["Sales"] = pd.to_numeric(
    df["Sales"],
    errors="coerce"
)
```

# pd.to_datetime()

```python
# pd.to_datetime() is used to convert a column
# into Pandas datetime format.

# It converts date values from string/object
# into datetime format.

# Syntax:
# df["Date"] = pd.to_datetime(df["Date"])

# DATE CONVERSION
#
# pd.to_datetime()
# → String/Object date ko datetime mein convert karta hai.
#
# errors="coerce"
# → Invalid date ko NaT bana deta hai.
#
# dayfirst=True
# → Date mein day ko first value maana jata hai.
#
# .dt.year
# → Year
#
# .dt.month
# → Month
#
# .dt.day
# → Day
#
# NaT
# → Not a Time / missing datetime value
```

```python
import pandas as pd

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],

    "Customer": [
        "Amit", "Rahul", "Neha", "Pooja", "Rohit",
        "Sneha", "Vikas", "Kiran", "Ankit", "Priya"
    ],

    "Order_Date": [
        "2025-01-15",
        "2025-02-20",
        "2025-03-10",
        "2025-04-25",
        "2025-05-18",
        "Invalid",
        "2025-07-12",
        "2025-08-30",
        "Invalid",
        "2025-10-15"
    ],

    "Joining_Date": [
        "15-01-2023",
        "20-02-2022",
        "10-03-2024",
        "25-04-2021",
        "18-05-2023",
        "12-06-2022",
        "Invalid",
        "30-08-2020",
        "15-09-2021",
        "10-10-2023"
    ],

    "Delivery_Date": [
        "2025/01/20",
        "2025/02/25",
        "2025/03/15",
        "Invalid",
        "2025/05/25",
        "2025/06/18",
        "2025/07/17",
        "Invalid",
        "2025/09/10",
        "2025/10/20"
    ]
}

df = pd.DataFrame(data)
df
```

```python
# ============================================================
# Q1
# ============================================================

# Order_Date column ko datetime format mein convert karo.

df["new_order_date"]=pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)
df
# ============================================================
# Q3
# ============================================================

# Delivery_Date column ko datetime mein convert karo.

df["new_delivery_date"]=pd.to_datetime(
    df["Delivery_Date"],
    errors="coerce"
)
df
# ============================================================
# Q4
# ============================================================

# Order_Date mein invalid values ko NaT mein convert karo.
# errors="coerce" ka use karo.
df["new_order_date"]=pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)
df

# ============================================================
# Q5
# ============================================================

# Order_Date se Year ka ek new column create karo.
df["year"]=df["new_order_date"].dt.year
df
df["year"].mean()
df["year"]=df["year"].fillna(df["year"].mean())
df
# ============================================================
# Q6
# ============================================================

# Order_Date se Month ka ek new column create karo.
df["month"]=df["new_order_date"].dt.month
df
df["month"].mean()
df["month"]=df["month"].fillna(df["month"].mean())
df
# ============================================================
# Q7
# ============================================================

# Order_Date se Day ka ek new column create karo.
df["day"]=df["new_order_date"].dt.day
df
df["day"].mean()
df["day"]=df["day"].fillna(df["day"].mean())
df

# ============================================================
# Q8
# ============================================================

# Order_Date se Month_Name ka new column create karo.

df["month_name"]=df["new_order_date"].dt.month_name()
df
df["month_name"].mode()
df["month_name"]=df["month_name"].fillna(df["month_name"].mode()[0])
df
# ============================================================
# Q9
# ============================================================

# Order_Date se Day_Name ka new column create karo.
df["day_name"]=df["new_order_date"].dt.day_name()
df
df["day_name"].mode()
df["day_name"]=df["day_name"].fillna(df["day_name"].mode()[0])
df
```
