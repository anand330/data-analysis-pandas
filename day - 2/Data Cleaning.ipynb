# Missing Values kya hain?

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Amit", "Neha", "Priya"],
    "Age": [25, 28, None, 30],
    "Salary": [35000, None, 45000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Missing Values
#
# Missing value means a value is not available in the dataset.
#
# Missing values are very common in real-world datasets.
#
# Common representations:
# NaN  -> Not a Number
# None -> Python null value
# NaT  -> Missing value for date/time data
#
# Example:
# Age = [25, 28, None, 30]
#
# Here, None represents a missing value.
#
# Why missing values are important:
# 1. They can affect calculations.
# 2. They can affect average and other statistics.
# 3. They can create problems during analysis.
# 4. They need to be identified and handled properly.
#
# Pandas provides different functions to identify and handle
# missing values.
```

# isnull()

```python
# isnull()
#
# isnull() is used to identify missing values.
#
# It returns:
# True  -> value is missing
# False -> value is not missing
#
# Check complete DataFrame:
# df.isnull()
#
# Check a specific column:
# df["Age"].isnull()
#
# Find rows where Age is missing:
# df[df["Age"].isnull()]
#
# isnull() is mainly used during data cleaning
# to identify missing values.
```

```python
print(df)

print(df.isnull()) 
print(df["Age"].isnull())  # Kisi particular column par isnull()

print(df[df["Age"].isnull()]) # Missing rows ko identify karna 
#
```

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Amit", "Rahul", "Neha", "Priya", "Rohit",
             "Pooja", "Ankit", "Sneha", "Vikas", "Kiran"],
    "Age": [25, None, 29, 32, None, 27, 30, None, 26, 31],
    "Department": ["IT", "HR", None, "Finance", "IT",
                   "Sales", None, "HR", "IT", "Finance"],
    "Salary": [35000, 42000, None, 55000, 48000,
               None, 60000, 39000, None, 52000],
    "Experience": [2, 4, None, 8, 6, 3, None, 5, 2, None],
    "City": ["Delhi", "Jaipur", "Mumbai", None, "Pune",
             "Delhi", None, "Jaipur", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df)
print(df.isnull())
print(df[df["Age"].isnull()])
print(df["Salary"].isnull())
```

# isna()

```python
# isna()
#
# isna() is used to identify missing values.
#
# True  -> value is missing
# False -> value is not missing
#
# Check complete DataFrame:
# df.isna()
#
# Check one column:
# df["Age"].isna()
#
# Find rows where Age is missing:
# df[df["Age"].isna()]
#
# isna() and isnull() have the same functionality.
```

```python
print(df)
print(df.isna())
print(df["Department"].isna())
print(df[df["Salary"].isna()])
```

# notnull()

```python
# notnull()
#
# notnull() is used to identify non-missing values.
#
# True  -> value is available
# False -> value is missing
#
# Check complete DataFrame:
# df.notnull()
#
# Check one column:
# df["Age"].notnull()
#
# Find rows where Age is available:
# df[df["Age"].notnull()]
#
# Find rows where Age and Salary are both available:
# df[df["Age"].notnull() & df["Salary"].notnull()]
```

```python
print(df)
print(df.notnull())   #Complete DataFrame
print(df["Salary"].notnull())   # Particular column
print(df["City"].notnull())  # Sirf valid rows identify karna

print(df[df["Salary"].notnull() & df["City"].notnull()])  # Multiple conditions
```

```python
import pandas as pd

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "Customer": ["Amit", "Rahul", "Neha", "Pooja", "Rohit",
                 "Ankit", "Sneha", "Vikas", "Kiran", "Priya"],
    "Product": ["Laptop", "Mobile", "Tablet", None, "Laptop",
                "Mobile", None, "Headphone", "Tablet", "Laptop"],
    "Quantity": [2, 1, None, 3, 1, None, 2, 4, 1, None],
    "Price": [55000, 25000, 30000, 15000, None,
              22000, 18000, None, 28000, 60000],
    "City": ["Delhi", "Jaipur", None, "Mumbai", "Pune",
             "Delhi", "Jaipur", None, "Mumbai", "Pune"]
}

df_sales = pd.DataFrame(data)
print(df_sales)
```

```python
print(df_sales)
#Q1.Identify karo ki Price column mein kaun-kaun se values available hain.
print(df_sales["Price"].notnull())

# Q2. Un orders ki complete rows display karo jinki Product information available hai.
print(df_sales[df_sales["Product"].notnull()])

# Q3. Un orders ko identify karo jinki Quantity aur Price dono available hain.
print(df_sales[df_sales["Quantity"].notnull()& df_sales["Price"].notnull()])


# Q4. Un orders ki complete rows display karo jinki City information available hai.

print(df_sales[df_sales["City"].notnull()])

# Q5. ⭐ Real-Life Data Analyst Question
# Sales analysis ke liye un orders ko identify karo jahan Product, Quantity aur Price — teeno information available hain.
print(df_sales[df_sales["Quantity"].notnull() & df_sales["Product"].notnull() & df_sales["Price"].notnull()])
```

 # Missing Values Count

```python
# Missing Values Count
#
# To count missing values in each column:
# df.isnull().sum()
#
# isnull() identifies missing values.
# sum() counts the True values.
#
# To count missing values in one column:
# df["Age"].isnull().sum()
#
# Missing values percentage:
# df.isnull().mean() * 100
#
# Missing value count and percentage are very important
# during data cleaning and data quality analysis.
```

```python
import  pandas as pd
```

```python
# Q1. Har column mein missing values count karo
df_sales.isnull().sum()

# Q2. City column mein missing values count karo
df_sales["City"].isnull().sum()

# Q3. Product column mein missing values count karo
df_sales["Product"].isnull().sum()

# Q4. Price column mein missing values count karo
df_sales["Price"].isnull().sum()

# Q5. Quantity column mein missing values count karo
df_sales["Quantity"].isnull().sum()
```

# dropna()

```python
# dropna()
#
# dropna() is used to remove missing values.
#
# By default, dropna() removes rows containing
# at least one missing value.
#
# Remove rows with missing values:
# df.dropna()
#
# Remove rows based on specific columns:
# df.dropna(subset=["Price"])
#
# Remove columns containing missing values:
# df.dropna(axis=1)
#
# axis=0 -> rows
# axis=1 -> columns
#
# how="all" removes a row only when
# all values in that row are missing.
```

```python
# Original DataFrame
df_sales

# Remove rows containing any missing value
df_sales.dropna()

# Remove rows where Price is missing
df_sales.dropna(subset=["Price"])

# Remove missing values from City column
df_sales["City"].dropna()

# Remove rows where City is missing
df_sales.dropna(subset=["City"])

# Remove columns containing missing values
df_sales.dropna(axis=1)

# Remove rows containing missing values
df_sales.dropna(axis=0)
```

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                    111, 112, 113, 114, 115],

    "Name": ["Amit", "Rahul", "Neha", "Pooja", "Rohit",
             "Sneha", "Vikas", "Kiran", "Ankit", "Priya",
             "Mohit", "Simran", "Raj", "Nisha", "Arjun"],

    "Age": [25, None, 29, 32, None, 27, 30, None, 26, 31,
            None, 28, 34, None, 29],

    "Department": ["IT", "HR", None, "Finance", "IT",
                   "Sales", None, "HR", "IT", None,
                   "Sales", "Finance", "IT", None, "HR"],

    "Salary": [35000, 42000, None, 55000, 48000,
               None, 60000, 39000, None, 52000,
               45000, None, 65000, 41000, None],

    "Experience": [2, 4, None, 8, 6, 3, None, 5,
                   2, None, 4, 7, None, 3, 5],

    "City": ["Delhi", "Jaipur", "Mumbai", None, "Pune",
             "Delhi", None, "Jaipur", "Mumbai", "Pune",
             None, "Delhi", "Jaipur", None, "Mumbai"]
}

df = pd.DataFrame(data)

df

 # 🔹 Part 1 — isnull()

# Q1  Complete DataFrame mein missing values identify karo using isnull().
df.isnull()

# Q2  Age column mein missing values identify karo using isnull().

df['Age'].isnull()

# Q3  Salary column mein missing values identify karo.

df['Salary'].isnull()

# Q4 Un employees ki complete rows display karo jinki Age missing hai.

df[df['Age'].isnull()]

# Q5 Un employees ki complete rows display karo jinki Department missing hai.

df[df["Department"].isnull()]
```

```python
## 🔹 Part 2 — isna()
# Q6 Experience column mein missing values identify karo using isna().

df["Experience"].isna()

# Q7 Un employees ki complete rows display karo jinki City missing hai.

df[df["City"].isna()]

#  Q8 Aise employees identify karo jinki Age OR Salary missing hai.

df[(df["Age"].isna()) | (df["Salary"].isna())]

```

```python
# 🔹 Part 3 — notnull()
# Q9 Un employees ki complete rows display karo jinki Salary available hai.

df[df["Salary"].notnull()]

# Q10  Un employees ki complete rows display karo jinka Department available hai.
df[df["Department"].notnull()]

#Q11 Aise employees identify karo jinki Age AND Experience dono available hain.
# Hint: &
df[(df["Age"].notnull() ) & (df["Experience"].notnull())]
```

```python
# 🔹 Part 4 — notna()
# Q12 City mein available values identify karo using notna().

df[df["City"].notna()]

# Q13 Un employees ki complete rows display karo jinki Salary available hai using notna().
df[df["Salary"].notna()]

# Q14 ⭐  Aise employees identify karo jinki Age, Salary aur City teeno available hain.

df[(df["Age"].notna()) & (df["Salary"].notna()) & (df["City"].notna()) ]
```

```python
# 🔹 Part 5 — Missing Values Count
# Q15 Har column mein missing values ka count nikalo.

df.isnull().sum()

# Q16 Sirf Age column mein missing values ka count nikalo.

df["Age"].isnull().sum()

# Q17 Sirf Salary column mein missing values ka count nikalo.

df["Salary"].isnull().sum()

# Q18 ⭐ Har column ke missing values ka percentage calculate karo.

df.isnull().mean() * 100
```

```python
# 🔹 Part 6 — dropna()
# Q19 Aisi complete rows remove karo jahan kisi bhi column mein missing value hai.

df.dropna()

#  Q20 ⭐ Real-Life Data Analyst Question
# Employee analysis ke liye sirf un records ko retain karo jahan:
# Age available ho
# Salary available ho
# Department available ho Baaki rows remove karo using dropna().

df[(df["Age"].notnull()) & (df["Salary"].notnull()) & (df["Department"].notnull())].dropna()
```
