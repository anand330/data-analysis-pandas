# fillna()

```python
# ==========================================
# fillna()
# ==========================================

# fillna() is used to replace missing values.

# Basic syntax:
# df["Column"].fillna(value)

# Text column:
# df["City"] = df["City"].fillna("Unknown")

# Numerical column:
# df["Quantity"] = df["Quantity"].fillna(0)

# Multiple columns:
# df = df.fillna({
#     "City": "Unknown",
#     "Department": "Unknown",
#     "Quantity": 0
# })

# fillna() normally returns a new result.
# To save the changes, assign the result back to df.

# Example:
# df["City"] = df["City"].fillna("Unknown")
```

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    
    "Name": ["Amit", "Rahul", "Neha", "Pooja", "Rohit",
             "Sneha", "Vikas", "Kiran", "Ankit", "Priya"],
    
    "Department": ["IT", None, "HR", "Finance", None,
                   "Sales", "IT", None, "HR", "Finance"],
    
    "City": ["Delhi", "Jaipur", None, "Mumbai", "Pune",
             None, "Delhi", "Jaipur", None, "Pune"],
    
    "Age": [25, None, 29, 32, None, 27, 30, None, 26, 31],
    
    "Salary": [35000, 42000, None, 55000, 48000,
               None, 60000, 39000, None, 52000],
    
    "Experience": [2, 4, None, 8, 6, None, 10, 5, 2, None],
    
    "Bonus": [5000, None, 7000, 10000, None,
              6000, 12000, None, 5000, 8000]
}

df = pd.DataFrame(data)
df
# Fill all missing values in the DataFrame with 0
df.fillna(0)

# Fill all missing values in the DataFrame with "anand"
df.fillna("anand")

# Fill all missing values in the DataFrame with 10
df.fillna(10)
```

```python
# ==========================================
# Fill Missing Values with Mean
# ==========================================

# Mean = Average

# Syntax:
# df["Column"].mean()

# Fill missing values with mean:
# df["Column"] = df["Column"].fillna(
#     df["Column"].mean()
# )

# Example:
# df["Salary"] = df["Salary"].fillna(
#     df["Salary"].mean()
# )

# Mean is generally used for numerical columns.

# Mean Formula:
# Mean = Sum of values / Number of values
```

```python
# Check the DataFrame
df

# Calculate mean of Salary
df["Salary"].mean()

# Fill missing Salary values with Salary mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Check the DataFrame
df


# Calculate mean of Age
df["Age"].mean()

# Fill missing Age values with Age mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Check the DataFrame
df
```

```python
# ==========================================
# Fill Missing Values with Median
# ==========================================

# Median is the middle value after sorting the data.

# Calculate median:
# df["Column"].median()

# Fill missing values with median:
# df["Column"] = df["Column"].fillna(
#     df["Column"].median()
# )

# Example:
# df["Salary"] = df["Salary"].fillna(
#     df["Salary"].median()
# )

# Median is useful when data contains outliers.
```

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    
    "Name": ["Amit", "Rahul", "Neha", "Pooja", "Rohit",
             "Sneha", "Vikas", "Kiran", "Ankit", "Priya"],
    
    "Age": [25, 28, None, 32, 27, None, 30, 26, None, 29],
    
    "Salary": [35000, 42000, None, 55000, 48000,
               None, 150000, 39000, None, 52000],
    
    "Experience": [2, 4, 3, None, 6, 5, None, 2, 4, None],
    
    "Performance_Score": [80, 85, None, 90, 88,
                          None, 95, 82, None, 91]
}

# Create DataFrame
df_median = pd.DataFrame(data)

# View DataFrame
df_median


# Calculate median of Salary
df_median["Salary"].median()

# Fill missing Salary values with Salary median
df_median["Salary"] = df_median["Salary"].fillna(
    df_median["Salary"].median()
)

# View DataFrame
df_median


# Calculate median of Age
df_median["Age"].median()

# Fill missing Age values with Age median
df_median["Age"] = df_median["Age"].fillna(
    df_median["Age"].median()
)

# View DataFrame
df_median
```

# Forward Fill (ffill())

```python
# Forward Fill (ffill) is used to fill missing values
# using the previous available value.

# Syntax:
# df["Column"] = df["Column"].ffill()

# Example:
# If values are:
# Delhi
# NaN
# NaN
# Jaipur

# After ffill:
# Delhi
# Delhi
# Delhi
# Jaipur
```

```python
import pandas as pd

data = {
    "Employee": ["Amit", "Rahul", "Neha", "Pooja", "Rohit"],
    "City": ["Delhi", None, None, "Jaipur", None]
}

df = pd.DataFrame(data)

# Fill missing City values with the previous available value
df["City"] = df["City"].ffill()
df
```

```python
# ffill() = Forward Fill
# It uses the previous value to fill NaN.
# It works from TOP → BOTTOM.
# It is useful for sequential/time-series data.
# The first missing value remains missing if there is no previous value.
```

# Backward Fill (bfill())

```python
# Backward Fill (bfill) is used to fill missing values
# using the next available value.

# Syntax:
# df["Column"] = df["Column"].bfill()

# Example:
# Before:
# Delhi
# NaN
# NaN
# Jaipur

# After bfill:
# Delhi
# Jaipur
# Jaipur
# Jaipur
```

```python
import pandas as pd

data = {
    "Employee": ["Amit", "Rahul", "Neha", "Pooja", "Rohit"],
    "City": [None, None, "Delhi", "Jaipur", None]
}

df = pd.DataFrame(data)

# Fill missing City values using the next available value
df["City"] = df["City"].bfill()
df
```

```python
# bfill() = Backward Fill
# It uses the next available value to fill NaN.
# It works from BOTTOM → UP.
# It is useful for sequential/time-series data.
# If there is no value after a missing value, it remains missing.
```

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Employee_Name": [
        "Amit", "Rahul", "Neha", "Pooja", "Rohit",
        "Sneha", "Vikas", "Kiran", "Ankit", "Priya"
    ],
    "Department": [
        "IT", None, None, "HR", None,
        "Finance", None, "IT", None, None
    ],
    "Shift": [
        "Morning", None, None, "Evening", None,
        "Night", None, "Morning", None, "Evening"
    ],
    "Manager": [
        None, None, "Raj", None, None,
        "Simran", None, None, "Aman", None
    ]
}

df = pd.DataFrame(data)
df

#Q1. Department column ke missing values ko previous available value se fill karo using ffill().

df["Department"].ffill()
df["Department"]=df["Department"].ffill()
df

# Q2. Shift column ke missing values ko next available value se fill karo using bfill().

df["Shift"]=df["Shift"].bfill()
df

# Q3. Manager column mein missing values ko ffill() se fill karo.
df["Manager"]=df["Manager"].ffill()
df

# Q4. Department column mein pehle ffill() aur phir bfill() apply karo, taaki maximum missing values fill ho saken.

df["Department"]=df["Department"].ffill()
df

df["Department"]=df["Department"].bfill()
df
```
