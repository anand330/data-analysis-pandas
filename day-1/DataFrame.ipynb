 # DATAFRAME
  1. DataFrame is a two-dimensional data structure.
  2. . It contains rows and columns.
  3. A DataFrame looks like a table.
  4. Each column of a DataFrame is a Series.
  5. DataFrame is one of the most important structures in Pandas.

```python
import pandas as pd
data={
    "name":['anand','rahul','amit'],
    "age":[23,45,56],
    "class":[9,8,12]
}
df=pd.DataFrame(data)
print(df)

```

```python
data2={
    "first_name":['anand','kiran','rahul'],
    "age":[23,44,21],
    "departments":['it','hr','it']
    }
# 1. Dictionary keys become column names.
# 2. Dictionary values become column data.
# 3. pd.DataFrame() converts the dictionary into a DataFrame.
# 4. The length of values should normally be the same
#    for all columns.
df2=pd.DataFrame(data2)
print(df2)

print(df2.size)
print(df2.ndim)
print(df2.shape)
```

```python
data=pd.DataFrame({
    'name':['anand','rahul','naveen'],
    'age':[23,34,56]
})
print(data)
print(data['age'])
print(data['name'])
```

```python
# Dictionary se DataFrame
data = {
    "Name": ["Amit", "Rahul", "Neha"],
    "Age": [25, 28, 24],
    "City": ["Delhi", "Jaipur", "Mumbai"]
}

df = pd.DataFrame(data)

print(df)
print(df['Name'])
print(df['Age'])
print(df['City'])
print(df.ndim)
print(df.size)
print(df.shape)
```

```python
data=[
    ['amit','rahul','mohit'],
    [34,45,23],
    ['jaipur','delhi','ajmer']
]
df=pd.DataFrame(
    data,
    columns=['name','age','city']
)
print(df)
# 1. A list of lists can be converted into a DataFrame.
# 2. Each inner list represents one row.
# 3. The columns parameter is used to give column names.
# 4. The number of column names should match
#    the number of values in each row.
```
