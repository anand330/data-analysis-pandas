```python
import pandas as pd

data = {
    "Name": ["Rahul", "Naveen", "Mohit", "Anand", "Neha"],
    "Age": [23, 28, 25, 32, 24],
    "City": ["Jaipur", "Delhi", "Ajmer", "Jaipur", "Mumbai"],
    "Department": ["IT", "HR", "Sales", "IT", "Finance"],
    "Salary": [35000, 42000, 30000, 55000, 38000]
}

df = pd.DataFrame(data)

df
```

```python
# 1. Single Column Select Karna
print(df['Name'])
print(df['Age'])
print(df['City'])
print(df['Department'])
```

```python
# 1. Single Column Select Karna
print(df['Name'])
print(df['Age'])
print(df['City'])
print(df['Department'])
```

```python
# 2. Multiple Columns Select Karna
df
print(df[['Name','Age']])
df[['Department','Age']]
df[['Name',"Age","Department"]]
```

```python
df["Name"]                 # Single column
df[["Name", "Salary"]]     # Multiple columns
```

```python
# 3. Column ko Position se Select Karna
# .iloc[] se column ki position ke according select kar sakte hain.
# df.iloc[:, 0]
# :  → All rows
# 0  → First column
print(df)
print(df.iloc[: , 2])
print(df.iloc[: , 3])
print(df.iloc[: , 0])
print(df.iloc[: , 1])
```

```python
#4. Multiple Columns Position se

print(df)
print(df.iloc[: , 0:2])
print(df.iloc[:, 1:3])
print(df.iloc[: , 0:3])
```

```python
# 5. .loc[] se Column Select
#  Column name ke according select karna:
# Label/name based selection.
print(df)
print(df.loc[: , 'Name'])
print(df.loc[: , ['Name','City']])
```

# Selecting Rows
1. First Row Select Karna :   df.iloc[0]
2. Second Row Select Karna :   df.iloc[1]
3. 0 → First row, 
   1 → Second row , 
   2 → Third row , 
   3 → Fourth row , 
   4 → Fifth row

4.  Multiple Rows Select Karna  : df.iloc[0:3]
5.  Last Row Select Karna
6.  .loc[] Se Row Select Karna 

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Naveen", "Mohit", "Anand", "Neha"],
    "Age": [23, 28, 25, 32, 24],
    "City": ["Jaipur", "Delhi", "Ajmer", "Jaipur", "Mumbai"],
    "Department": ["IT", "HR", "Sales", "IT", "Finance"],
    "Salary": [35000, 42000, 30000, 55000, 38000]
}

df = pd.DataFrame(data)

print(df)
print(df.iloc[0])  #0 first row ka position hai.
print(df.iloc[2])  

print(df.iloc[0:2]) # Multiple Rows Select Karna
print(df.iloc[0:4]) # 0,1,2,3 rows 

print(df.loc[0]) 
print(df.loc[0:3]) # → 0, 1, 2 , 3 rows 



```

```python
 # Condition ke Basis Par Row Select Karna
print(df)
print(df[df["Age"]>30])

print(df[df["Salary"]> 35000])
print(df[df['City'] == 'Jaipur'])
```
