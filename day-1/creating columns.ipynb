 #  TOPIC 8: STRING SE NEW COLUMN
  Common operations:
   1 .str.upper()
   
2   .str.lower()

3 .str.title()

4 .str.strip()

5 .str.split()

6 .str.len()

```python
import pandas as pd

data = {
    "Name": ["Rahul Verma", "Naveen Kumar", "Mohit Sharma", "Anand Singh"],
    "City": ["Jaipur", "Delhi", "Ajmer", "Mumbai"]
}

df = pd.DataFrame(data)

df
```

```python
# 1. Original DataFrame ko print karna
print(df)

# 2. String ko UPPERCASE mein convert karke new column banana
# .str.upper()
df["new_name"] = df["Name"].str.upper()
print(df)

# 3. String ko lowercase mein convert karke new column banana
# .str.lower()
df["new_name2"] = df["Name"].str.lower()
print(df)

# 4. String ko Title Case mein convert karke new column banana
# .str.title()
df["df_name3"] = df["Name"].str.title()
print(df)

# 5. String ko words/list mein split karna
# .str.split()
df["new_name4"] = df["Name"].str.split()
print(df)

# 6. Split ke baad first word select karna
# .str.split().str[0]
# Example: "Rahul Verma" → "Rahul"
df["new_name5"] = df["Name"].str.split().str[0]
print(df)

# 7. String ki length nikalna
# .str.len()
df["length"] = df["Name"].str.len()
print(df)
```

1. .str.upper()       → UPPERCASE
2. .str.lower()       → lowercase
3. .str.title()       → Title Case
4. .str.split()       → String ko split karna
5. .str.split().str[0] → First word select karna
6. .str.len()         → String length

# Date & Time

```python
data = {
    "Name": ["Rahul", "Naveen", "Mohit"],
    "Joining_Date": ["2023-05-10", "2022-08-15", "2024-01-20"]
}

df = pd.DataFrame(data)

df
```

pd.to_datetime()

String ko proper datetime mein convert karne ke liye:

```python
df["Joining_Date"]=pd.to_datetime(df["Joining_Date"])
print(df)
```

String Date

    ↓
pd.to_datetime()

Datetime

```python
df["year"]=df["Joining_Date"].dt.year
print(df)

df["month"]=df["Joining_Date"].dt.month
print(df)

df["day"]=df["Joining_Date"].dt.day
print(df)

df["month_name"]=df["Joining_Date"].dt.month_name()
print(df)

df["day_name"]=df["Joining_Date"].dt.day_name()
print(df)
```

```python
data = {
    "Name": ["Rahul", "Naveen", "Mohit", "Anand"],
    "Joining_Date": [
        "2023-05-10",
        "2022-08-15",
        "2024-01-20",
        "2021-12-25"
    ]
}

df = pd.DataFrame(data)
print(df)

df["joining_date"]=pd.to_datetime(df["Joining_Date"])# String ko datetime mein convert karta hai.
print(df)


df["year"]=df["joining_date"].dt.year  #➡️ Year
print(df)

df['month']=df["joining_date"].dt.month  #➡️ Month number
print(df)

df["day"]=df["joining_date"].dt.day  #➡️ Day
print(df)

df["month_name"]=df["joining_date"].dt.month_name()   # ➡️ Month name
print(df)

df["day_name"]=df["joining_date"].dt.day_name()   #➡️ Day name
print(df)
```
