 # PANDAS SERIES

  1 A Series is a one-dimensional labeled data structure.

  2  A Series can store multiple values.

  3  Each value in a Series has an index.

  4  By default, Pandas gives numeric indexes starting from 0.

  5 We can create a Series using pd.Series().

  6 Syntax: pd.Series(data)


```python
import pandas as pd
data = pd.Series([10,12,23,34,45,64,45])
print(data)

data1=pd.Series([100,1000,10000,100000,100000])
print(data1)
```

```python
marks=pd.Series([80,90,75,60])
print(marks)
```

```python
names = pd.Series(['anand verma' , 'rahul' , 'mohit'])
print(names)
```

```python
marks=pd.Series([80,40,50,60])
print(marks)
# 0, 1, 2, 3 → Index
# 80, 75, 90, 85 → Values

print(marks[3])
print(marks[2])
```

```python
marks=pd.Series(
    [30,40,50,60,70],
    index=['math','science','english','computer','hindi']
)
print(marks)
print(marks['hindi'])
print(marks['english'])
print(marks['science'])
print(marks)
```

```python
import numpy as np
arr=np.arange(10,101,20)
data=pd.Series(arr)
print(data)
```

```python
import pandas as pd
marks=pd.Series(
    [100,20,30],
    index=['math','science','english']
)
print(marks)

import numpy as np
arr=np.arange(1,101,10)
data=pd.Series(arr)
print(data)

arr=np.random.randint(10,100,20)
data=pd.Series(arr)
print(data)
print(data[0:10])
```

```python
marks = pd.Series([80, 75, 90, 85])

print(marks.index)
print(marks.values)
print(marks.size)
print(marks.dtype)
print(marks.shape)
```
